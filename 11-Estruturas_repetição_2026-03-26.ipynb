## Instancie um objeto da classe lista contendo 5.000 valores inteiros iniciando em 0. 
Quantos são pares e quantos são impares?

## 01 – Descreva o racional da sua resposta:

R.: Primeiro foi criada uma lista usando o comando range para definir a quantidade de valores que no caso era 5 mil. Ja para contar foi utilizado o comando sum que é o comando de contagem e o para contar os pares se usa % 2 então para os pares foi isado a condicional de que se o x % 2 for == (igual) a 0 é par e se for != (diferente) de 0 é impar.

## 02 – Desenhe o fluxograma da resposta:

<img src="./fluxograma 11_2.png" width="400" />

## 03 – Escreva o programa em Python:

```python
lista_1 = list(range(5000))
pares = 0
impares = 0
i = 0
while i < len(lista_1):
    if lista_1[i] % 2 == 0:
        pares += 1
    else:
        impares += 1
    i += 1
print("Pares:", pares)
print("Ímpares:", impares)
```

## 04 – Substitua a estrutura de repetição e responda novamente a pergunta anterior.

```python
total = len(lista_1)
pares = total // 2
impares = total // 2
print("Pares:", pares)
print("Ímpares:", impares)
```

## Importe o .csv exportado na última aula e instancie um objeto da classe DataFrame.

Dos números de letras nos genes, quais são pares e quantos são impares?

## 05 – Descreva o racional da sua resposta:

R.: O arquivo CSV foi importado e convertido em DataFrame utilizando pandas. Em seguida, os nomes dos genes foram utilizados para calcular o número de caracteres de cada gene por meio do str.len(). Depois disso, os tamanhos foram divididos em pares ou ímpares utilizando a operação módulo % 2. Por fim, foram contabilizados e exibidos os genes com quantidade de caracteres pares e ímpares.

## 06 – Desenhe o fluxograma da resposta:

<img src="./conversão DataFrame Par_Impar.png" width="400" />

## 07 – Escreva o programa em Python:

```python
import pandas as pd

df = pd.read_csv('genes.csv', index_col=0)

tamanho_dos_genes = df.index.str.len()

pares = tamanho_dos_genes[tamanho_dos_genes % 2 == 0]
impares = tamanho_dos_genes[tamanho_dos_genes % 2 != 0]

print("Quantidade de pares:", len(pares))
print("Quantidade de ímpares:", len(impares))

print("Genes pares:", df.index[tamanho_dos_genes % 2 == 0])
print("Genes ímpares:", df.index[tamanho_dos_genes % 2 != 0])
```

## 08 – Substitua a estrutura de repetição e responda novamente a pergunta anterior.

```python
import pandas as pd
df = pd.read_csv('genes.csv', index_col=0)
i = 0
pares = 0
impares = 0
genes = list(df.index)
while i < len(genes):
    if len(genes[i]) % 2 == 0:
        pares += 1
    else:
        impares += 1
    i += 1
print("Quantidade de pares:", pares)
print("Quantidade de ímpares:", impares)
```
